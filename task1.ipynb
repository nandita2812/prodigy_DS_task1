## Dropping unnecessary columns from database


```python
df.drop(['Country Name'],axis=1,inplace=True)
```

## Some Important operation on data

```python
# filter the data for population

total_pop_data=df[df['Series Code']=='SP.POP.TOTL']
#sorting the data of 2021
total_population_sort=total_pop_data.sort_values(by ="2020",ascending=False)
# get top 10 countries with the highest total population for 2021
total_top_ten_countries=total_population_sort.head(10)

print("top 10 countries of total population \n:" )
print(total_top_ten_countries[['Country Code']])
```

## bar plot

### top 10 countries of total population in year 2020 and 2015

```python
# creating a Bar plot
plt.figure(figsize=(15,6))
plt.subplot(2,2,1)
sns.barplot(x="2020", y="Country Code", data=total_top_ten_countries, palette="coolwarm")
plt.title("top ten countries of total population(2020)",fontsize=10)
plt.xlabel("total population",fontsize=10)
plt.ylabel("country",fontsize=10)
plt.show()

# creating a Bar plot

plt.figure(figsize=(15,6))
plt.subplot(2,2,2)
sns.barplot(x="2015",y="Country Code", data=total_top_ten_countries, palette="coolwarm")
plt.title("top ten countries of total population(2015)",fontsize=10)
plt.xlabel("total population",fontsize=10)
plt.ylabel("country",fontsize=10)
plt.show()
```

## finding top 10 countries with highest male population

```python
male_population_data = df[df['Series Code']=='SP.POP.TOTL.MA.IN']
male_population_sort=male_population_data.sort_values(by='2020',ascending=True)
# sorting
male_ten_countries=male_population_sort.head(10)
print ('top 10 countries of male population')
print(male_ten_countries[['Country Code']])
```

## Finding top 10 countries with highest Female population

```python
Female_population_data = df[df['Series Code']=='SP.POP.TOTL.FE.IN']
Female_population_sort=Female_population_data.sort_values(by='2020',ascending=True)
# sorting
Female_ten_countries=male_population_sort.head(10)
print ('top 10 countries of Female population')
print(Female_ten_countries[['Country Code']])
```

```python
# creating a Bar plot
plt.figure(figsize=(15,6))
plt.subplot(2,2,1)
sns.barplot(x="2020", y="Country Code", data=male_ten_countries, palette="viridis")
plt.title("top 10 countries of male population(2020)",fontsize=10)
plt.xlabel("Male population",fontsize=10)
plt.ylabel("country",fontsize=10)
plt.show()

# creating a Bar plot

plt.figure(figsize=(15,6))
plt.subplot(2,2,2)
sns.barplot(x="2020",y="Country Code", data=Female_ten_countries, palette="viridis")
plt.title("top 10 countries of female population(2020)",fontsize=10)
plt.xlabel("Female population",fontsize=10)
plt.ylabel("country",fontsize=10)
plt.show()
```
